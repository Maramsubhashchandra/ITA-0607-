```python
import csv

data = [
    ["Sky","AirTemp","Humidity","Wind","Water","Forecast","EnjoySport"],
    ["Sunny","Warm","Normal","Strong","Warm","Same","Yes"],
    ["Sunny","Warm","High","Strong","Warm","Same","Yes"],
    ["Rainy","Cold","High","Strong","Warm","Change","No"],
    ["Sunny","Warm","High","Strong","Cool","Change","Yes"]
]

with open("training_data.csv", "w", newline="") as f:
    csv.writer(f).writerows(data)

print("CSV file created successfully!")
```

```python
import csv

# Read CSV
with open("training_data.csv") as f:
    data = list(csv.reader(f))

examples = data[1:]
n = len(data[0]) - 1

S = ['Ø'] * n
G = [['?'] * n]

def covers(h, x):
    return all(h[i] == '?' or h[i] == x[i] for i in range(n))

def more_general(h1, h2):
    return all(h1[i] == '?' or h1[i] == h2[i]
               for i in range(n))

for x, label in [(r[:-1], r[-1]) for r in examples]:

    if label == "Yes":

        # Remove inconsistent G
        G = [g for g in G if covers(g, x)]

        # Generalize S
        for i in range(n):
            if S[i] == 'Ø':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

    else:

        new_G = []

        for g in G:
            if covers(g, x):
                for i in range(n):
                    if g[i] == '?':
                        for value in set(r[i] for r in examples):
                            if value != x[i]:
                                h = g.copy()
                                h[i] = value

                                if more_general(h, S):
                                    new_G.append(h)
            else:
                new_G.append(g)

        G = new_G

# Remove duplicate hypotheses
G = list({tuple(g) for g in G})

print("Training Examples:")
for r in examples:
    print(r)

print("\nSpecific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")
for g in G:
    print(list(g))
```
