```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text

# Dataset
data = pd.DataFrame({
    'Outlook':['Sunny','Sunny','Overcast','Rain','Rain','Rain','Overcast','Sunny','Sunny','Rain','Sunny','Overcast','Overcast','Rain'],
    'Temperature':['Hot','Hot','Hot','Mild','Cool','Cool','Cool','Mild','Cool','Mild','Mild','Mild','Hot','Mild'],
    'Humidity':['High','High','High','High','Normal','Normal','Normal','High','Normal','Normal','Normal','High','Normal','High'],
    'Wind':['Weak','Strong','Weak','Weak','Weak','Strong','Strong','Weak','Weak','Weak','Strong','Strong','Weak','Strong'],
    'Play':['No','No','Yes','Yes','Yes','No','Yes','No','Yes','Yes','Yes','Yes','Yes','No']
})

X = pd.get_dummies(data.drop('Play', axis=1))
y = data['Play']

# Build ID3 tree using Entropy
model = DecisionTreeClassifier(criterion='entropy', random_state=0)
model.fit(X, y)

print("ID3 Decision Tree:")
print(export_text(model, feature_names=list(X.columns)))

# Classify a new sample
sample = pd.DataFrame([['Sunny','Cool','High','Strong']],
                      columns=['Outlook','Temperature','Humidity','Wind'])
sample = pd.get_dummies(sample).reindex(columns=X.columns, fill_value=0)

print("New Sample:", model.predict(sample)[0])
```
