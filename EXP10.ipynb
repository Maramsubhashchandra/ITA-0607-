```python
import numpy as np
from sklearn.mixture import GaussianMixture

# Dataset
X = np.array([[1],[2],[3],[4],[10],[11],[12],[13]])

# EM algorithm using Gaussian Mixture Model
em = GaussianMixture(n_components=2, random_state=0)
em.fit(X)

# Cluster prediction
labels = em.predict(X)

print("Data:", X.flatten())
print("Cluster Labels:", labels)
print("Means:", em.means_.flatten())
print("Weights:", em.weights_)
```
