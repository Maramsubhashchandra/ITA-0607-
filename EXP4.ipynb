```python
import numpy as np

# XOR dataset
X = np.array([[0,0],[0,1],[1,0],[1,1]])
Y = np.array([[0],[1],[1],[0]])

# Initialize weights and biases
np.random.seed(1)
W1 = np.random.randn(2,4)
b1 = np.zeros((1,4))
W2 = np.random.randn(4,1)
b2 = np.zeros((1,1))

lr = 0.5

def sigmoid(x):
    return 1/(1+np.exp(-x))

# Backpropagation training
for epoch in range(10000):
    # Forward propagation
    H = sigmoid(X @ W1 + b1)
    O = sigmoid(H @ W2 + b2)

    # Error
    error = Y - O

    # Backpropagation
    dO = error * O * (1-O)
    dH = (dO @ W2.T) * H * (1-H)

    # Update weights and biases
    W2 += lr * H.T @ dO
    b2 += lr * np.sum(dO, axis=0, keepdims=True)
    W1 += lr * X.T @ dH
    b1 += lr * np.sum(dH, axis=0, keepdims=True)

# Testing
pred = (O >= 0.5).astype(int)

print("Actual Output:")
print(Y.flatten())

print("\nPredicted Output:")
print(pred.flatten())

print("\nAccuracy:", np.mean(pred == Y) * 100, "%")
```
