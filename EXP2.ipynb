```python
import pandas as pd

data = pd.DataFrame([
['Sunny','Warm','Normal','Strong','Warm','Same','Yes'],
['Sunny','Warm','High','Strong','Warm','Same','Yes'],
['Rainy','Cold','High','Strong','Warm','Change','No'],
['Sunny','Warm','High','Strong','Cool','Change','Yes']
], columns=['Sky','Temp','Humidity','Wind','Water','Forecast','Enjoy'])

data.to_csv('training.csv', index=False)
print(data)
```

```python
attrs = list(data.columns[:-1])
values = [data[a].unique().tolist() for a in attrs]
S = ['0'] * len(attrs)
G = [['?'] * len(attrs)]

def covers(h, x):
    return all(a == '?' or a == v for a, v in zip(h, x))

for _, row in data.iterrows():
    x, y = row.iloc[:-1].tolist(), row.iloc[-1]
    if y == 'Yes':
        G = [g for g in G if covers(g, x)]
        for i in range(len(S)):
            if S[i] == '0': S[i] = x[i]
            elif S[i] != x[i]: S[i] = '?'
    else:
        newG = []
        for g in G:
            if covers(g, x):
                for i in range(len(attrs)):
                    if g[i] == '?':
                        for v in values[i]:
                            if v != x[i] and (S[i] == '?' or v == S[i]):
                                h = g.copy(); h[i] = v
                                newG.append(h)
            else:
                newG.append(g)
        G = newG

print("Specific Boundary (S):", S)
print("General Boundary (G):")
for g in G:
    print(g)
```
